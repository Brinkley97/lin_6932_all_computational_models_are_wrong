### True probabilities vs empirical probabilities

````python
import numpy as np
import pandas as pd
from matplotlib import pyplot as plt
````

Say I'm back to rolling the dice. The true probabilities of rolling each of 1, 2, 3, ..., 6 are stored in the variable `dice_probability`.

````python
dice_probability = np.array([.3, .2, .1, .2, .15, .05])
print(dice_probability)
print(dice_probability[0]) # the probability to roll 1
print(dice_probability[1]) # the probability to roll 2, etc.
````

````python
# Visualize the probability distribution as a bar plot
# It is also useful to visualize the distribution
def plot_dice_probability(p):
  x = np.arange(6) + 1
  plt.bar(x, p)
  plt.xlabel("Outcome x")
  plt.ylabel("Probability p(x)")

plot_dice_probability(dice_probability)
````

````python
# We can simulate random dice rolls in this way
x = np.arange(6) + 1 # why +1?
print(np.random.choice(x, p=dice_probability)) # Simultating one random dice roll
````

````python
# Let's simulate 100 dice rolls:
N = 100
sample = np.random.choice(x, p=dice_probability, size=10000)
sample
````

````python
# Now we can compute the empirical frequencies of the outcomes in the sample
from collections import Counter
sample_frequency = np.zeros(6)
for outcome in sample:
  sample_frequency[outcome-1] += 1 # why -1?
sample_frequency
````

````python
# And compare these to the underlying probabilities that were used to generate the data.
def double_bar(x, y1, y2, width=0.4):
  plt.bar(x-width/2, y1, width)
  plt.bar(x+width/2, y2, width)

N = 10000

double_bar(x, dice_probability, sample_frequency/N)
plt.xlabel("Outcome")
plt.ylabel("Probability")
plt.legend(["True probability", "Sample frequency"])
````

### A simple generator

When discussing probability, we talked about solving problems with probability. One type of problems is ***generation and prediction***. For instance, Just learning the probability of getting H or T isn't always terribly interesting. In cognitive modeling (and many other aspects of NLP), we are interested in what these probabilities can guide our comprehension, production, learning, and online prediction.

Here we will demonstrate the idea of generation with a simple example. Let's say I'm trying to produce some sequence of symbols based on the results of rolling a fair die. If I get 1 or 2, this generates the symbol 'this', if I get 3 or 4, this generates the symbo 'is', otherwise we get the symbol 'wug'. I roll this die to generate a sequence of 8 symbols. How many possible sequence could I have? Could you write a function that defines a generator that will generate ***one*** possible 8-symbol sequence randomly?

````python
### Insert your code here ###

def generator():
  sequence = []
  for i in range (8):
    outcome = random.choices([1,2,3,4,5,6], weights = [1/6, 1/6, 1/6, 1/6, 1/6, 1/6])
    if outcome[0] in [1,2]:
      sequence.append('this')
  return sequence

generator()
````

## Setting up the question

Let's say we are trying to distinguish/categorize two sounds, /b/ and /p/, based on their voice onset time (VOT) (what are some of the assumptions we are making here)?

From Bayesian inversion equation, we know that we can write the following equation, using *S* as the variable for sound category, and *VOT* as the variable for acoustic signal/representation (VOT in this case):

$P(S | VOT) = \frac{P(VOT | S)P(S)}{P(VOT)}$

## *Calculating* the Prior and the Posterior

In class, we discussed that we can esimate the prior of a certain sound category from corpus. But what if we don't have a corpus of this sort?

In a way, we can think of *prior* as the probability of a variable before learning further information (think about the literal meaning of the word "prior"). For instance, in this case we can think of $P(/b/)$ as the probability of the sound /b/ before learning ANYTHING about its acoustic representations/cudes. What might $P(/b/)$ be?

Now think about how you would calculate the posterior $P(/b/ | VOT)$ for **a series of VOT values**. Please take some time to check the code cells below for correctness and understand what they do.

````python
## Insert your code here
list_vot_values = list(np.arange(0, 100, 1))
````

````python
def likelihood_function(mu, sigma, VOT):
  likelihood = 1 / ((2 * math.pi * sigma ** 2) ** 0.5) * math.e ** (-1 * (VOT - mu) ** 2 / (2 * sigma ** 2))
  return likelihood
````

````python
posterior_values = []
for vot_value in list_vot_values:
  numerator = likelihood_function(30, 8, vot_value) * 0.75
  denominator = likelihood_function(30, 8, vot_value) * 0.75 + likelihood_function(100, 12, vot_value) * 0.25
  posterior = numerator / denominator
  posterior_values.append(posterior)
````

````python
sns.lineplot(x = list_vot_values, y = posterior_values)
````

Now we move on to getting bigrams from the full corpus. We will store the bigram counts in a python dictionary

````python
from collections import defaultdict #we need the defaultdictionary library for counting bigrams in the box below
model = defaultdict(lambda: defaultdict(lambda: 0))
print(model)
````

````python
for sentence in gutenberg.sents('shakespeare-caesar.txt'): #for all sentences
    #words should be lowercase (lower()), remove all punctuation (isalpha())
    words = [word.lower() for word in sentence if word.isalpha()]

    for w1, w2 in bigrams(words, pad_right=True, pad_left=True): #now we count bigrams according to their occurrence
        model[w1][w2] += 1
````
