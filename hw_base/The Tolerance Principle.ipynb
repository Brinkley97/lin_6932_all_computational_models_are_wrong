## The Tolerance Principle at a high level and a very high level

 Virtually all potential rules of a language are bound to have (arbitrary) exceptions, so the learner shouldn't hold rules to a standard of perfection. For example, a clear rule of English, suffix *-ed* to the verb to make past tense has plenty of exceptions. The real question is how many exceptions, as a proportion of the total number of types that are expected to follow the rule, can be tolerated. If there are “too many” exceptions, the rule is probably not a good one for the learner to acquire. If there aren’t “too many,” the learner should acquire the rule. The Tolerance Principle provides this threshold: If the rule should apply to $N$ types that the child knows so far, but the child has learned $e$ exceptions to the rule, the child should learn the rule + memorize the exceptions if $e$ < $\theta$, the tolerance threshold. If $e$ > $\theta$, it’s back to the drawing board. Either just memorize it all as exceptions, or try again with a different rule.

One one hand, language is incredibly arbitrary. On the other, it is ex-traordinarily rule-governed. This duality is what has secured linguistics its privileged role in the cognitive sciences. That said, it also presents an enormous challenge to language learners, who each have to work out for themselves what is productive/rule-governed and what is non- productive/arbitrary. The Tolerance Principle is a precise metric which the learner can use to make this decision.

**TP**: $\theta = \frac{N}{lnN}$

**The question**: Is $e < \theta$ ?

### Strong, Weak, Semi-Weak, and Suppletive

Four kinds of verbs are traditionally distinguished in traditional Germanic linguistics, strong verbs, weak verbs, semi-weak verbs, and suppletive verbs. Weak verbs are the ones that make pasts in -*ed* in English without any kind of stem change. These are regular in English. Strong verbs are ones that undergo a stem vowel change instead of suffixing to form pasts, like *sing*, *bite*, or *ride*. Semi-weak verbs both undergo stem change and suffixing a coronal obstruent (-*t* or -*d*) like sleep or tell. Finally, suppletives are just straight-up irregulars like *go*.
The terminology actually goes back to the Brothers Grimm. Mainly famous for their fairy tales, they were also foundational historical linguists and comparative Germanicists.
