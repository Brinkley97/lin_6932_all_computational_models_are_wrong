### Surprisal

Information theory is based on an equivalence between the idea of "surprise" and the idea of "information". A more surprising event contains more information. An unsurprising event contains little information. 
- > Example with Brittany (my girlfriend): Say in December, I attempt to surprise her. Either (and on the extremes so I can understand)
    - > a. She does not know (before hand): high surprisal because it was much surprise
    - > b. She does know (before hand): low surprisal because it wasn't too much of a surprise

The exact link is that the **information content** $h(x)$ of an event $x$ is:
$$
h(x) = -\log p(x).
$$
For a logarithm base 2, the unit of information is **bits**. For a natural logarithm, the unit of information is **nats**.

### Entropy

The next most important concept in information theory is **entropy**. Entropy quantifies uncertainty about an outcome before you have observed it. Surprisal quantifies how surprising an event is after you have observed it.

> Notice:
> | Concept   | Meaning     | Timing              |
> |-----------|-------------|---------------------|
> | Entropy   | Uncertainty | Before observed     |
> | Surprisal | Surprising  | After observed      |

We'll think about entropy by imagining some weighted dice.

> Weighted dice means not every outcome has the same chance of occuring.
Some are given more importance with a higher weight, thus leading to an increased chance of occuring.

Entropy is the **average expected surprise** for the dice roll. Let the set of possible outcomes from the dice roll be the set $\{1, 2,3,4,5,6\}$. And let $p(x)$ denote the probability of outcome $x$. Entropy is the average surprise of each possible outcome, weighted by its probability:
$$
\operatorname{H}[p] = -\sum_{x} p(x) \log p(x).
$$
Below, let's write a function for entropy and apply it to the dice probabilities. (The first function written here will not be the final function.)

```python
dice_probability = np.array([.3, .2, .1, .2, .15, .05])

def entropy_not_final(p): # input p is an array of probabilities, ie non-negative numbers that sum to 1
  return -np.sum(p * np.log(p))

entropy_not_final(dice_probability)
```

So the entropy of the probability distribution defined in `dice_probability` is around 1.7 nats. How about this other distribution?

```python
new_dice_probability = np.array([.3, .2, .1, .2, .2, 0])

def plot_dice_probability(p):
  x = np.arange(6) + 1
  plt.bar(x, p)
  plt.xlabel("Outcome x")
  plt.ylabel("Probability p(x)")

plot_dice_probability(new_dice_probability)
```

We can't compute this because one of the outcomes has probability 0. The entropy contains a term $0 \log 0$, which `numpy` can't calculate. In information theory, we always set
$$
0 \log 0 = 0
$$
so we need to use a special logarithm function that reflects this.

```python
import scipy.special

def entropy(p):
  return -np.sum(scipy.special.xlogy(p, p)) # scipy.special.xlogy(x,y) computes x*log(y) with special case 0*log(0) = 0.

entropy(new_dice_probability)
```

### Cross Entropy

Entropy quantifies your uncertainty about the outcome of a random event like a dice roll, *assuming you know the true probabilities of all the outcomes*. What if you don't know the probabilities of the outcomes? Or what if you *think* you know the probabilities of the outcomes, but you're wrong? For example, what if you think it's a fair die (with probabilities 1/6 for each outcome), but it's actually weighted?

Cross entropy allows us to think about uncertainty when your *subjective* beliefs about probabilities might differ from the *objectively* true probabilities. Let $p(x)$ be the true probability for an event $x$, and $q(x)$ be its subjective probability. The cross entropy of $p$ and $q$ is
$$
\operatorname{H}[p,q] = -\sum_{x \in \mathcal{X}} p(x) \log q(x).
$$
This is your average surprise if you *think* the probabilities are $q(x)$, but they are *actually* $p(x)$. First, make sure the formula makes sense to you. Then let's see some examples.

> The weight is $p(x)$ as this will either amplify our *subjective* beliefs or dismantle.

> To amplify, I believe $p(x)$ will tend to 1 because it is multiplying by our $ \log q(x) $, where multiplication in this case means to increase how important the subject belief is.
- > p(x) \log q(x) $ \Rightarrow $ 1 * .7 = .7, where this is support.
> To dismantle, I believe $p(x)$ will tend to 0 because it is multiplying by our $ \log q(x) $, where multiplication in this case means to decrease how important the subject belief is.
- > $ p(x) \log q(x) $  $ \Rightarrow $ 0 * .7 = 0, where this is no support.

**Exercise** Let's say you think someone is rolling fair dice, so $p(x) = 1/6$. But actually the dice follow a weighted distribution `dice_probability` as above. What is the cross entropy?

## KL Divergence

The last information theory quantity we'll need for now is **Kullback-Leibler (KL) divergence**, which is just the difference between entropy and cross entropy for some distributions $p(x)$ and $q(x)$:

$$
D_{\text{KL}}[p || q] = H[p,q] - H[p].
$$

KL divergence is important because it measures a distance from one probability distribution $p$ to another probability distribution $q$. It can be interpreted as the *amount of extra surprise* that you experience due to having the wrong subjective probability distribution $q$.

> Extra surprise is when H[p,q] < H[p] or H[p,q] > H[p]?

> $ D_{\text{KL}}[p || q] = 0 \rightarrow $ no surprise, so no extra surprise

> $ D_{\text{KL}}[p || q] = 1 \rightarrow $ all surprise, so extra surprise

## Using cross entropy to fit a model to data

We will use our information-theoretic concepts to fit a model to data. We will fit a Zipf Distribution to the word frequency data.

You can read more about the Zipf Distribution here: https://en.wikipedia.org/wiki/Zipf%27s_law

G. K. Zipf (1935) noticed that the probability distribution of single words chosen at random from a text can be well-described with the following equation:
$$
p(w) = \frac{1}{Z} k^{-\alpha},
$$
where:
* $Z$ is the normalizing constant that makes the distribution sum to 1.
* $k$ is the frequency rank of the word $w$. If $w$ is the most frequent word, it has frequency rank $k=1$, etc.
* $\alpha$ is a nonnegative number.

The Zipf distribution is implemented below for a vocabulary of size $N$:

```python
def zipf_distribution(N, alpha):
  k = np.arange(N).astype(float) + 1 # an array of [1, 2, ..., N]
  p = k ** -alpha
  Z = p.sum()
  return p / Z
```

Now we want to **fit the model** consisting of the Zipf distribution to our frequency data. That means we want to find parameters for the Zipf distribution to make it match the frequency data as closely as possible. In this case, there is only one parameter, $\alpha$. In general, the way we do this is to *find parameters that minimize cross entropy between the data and the model.*

More specifiaclly, if the data has distribution $p(x)$, and the model instantiates a family of distributions $q(x)$, we want to find the specific $q(x)$ that minimizes $H[p,q]$.

In our case, the model is a Zipf distribution with parameter $\alpha$. So we want to find $\alpha$ to minimize the cross entropy between the empirical frequncies $p$ and the Zipf probabilities $q$ with parameter $\alpha$.

Below, we will show a generic way to do this minimization to fit the model. In doing so we will demonstrate that this procedure makes sense by performing a simple check: if we have data that we *know* was generated from a Zipf distribution with some particular $\alpha$, can we recover that same $\alpha$ from the data using our model fitting procedure?

```python
# Generate some data that we will use to validate our model-fitting procedure
N = 10000
true_alpha = 0.6538
data = zipf_distribution(10000, true_alpha)
data
```

We want to find parameter $\alpha$ to minimize cross entropy between the data $p$ and the model $q$ which is a function of $\alpha$. To do that, we need an **objective function**: a function from the parameter $\alpha$ to the cross-entropy of the data:

```python
def objective(parameters):
  alpha = parameters[0]
  return cross_entropy(data, zipf_distribution(N, alpha))

objective([.2]) # Cross entropy when Zipf alpha=.2
```

Now we will find the value of $\alpha$ to minimize the objective function using `scipy.optimize.minimize`:

```python
import scipy.optimize
result = scipy.optimize.minimize(objective, x0=np.array([1])) # x0 is an array of initial guesses for the parameters
zipf_fitted_alpha = result.x[0]
zipf_fitted_alpha
```

This shows that our fitting procedure can recover the true Zipf parameter when it is known. That means that our estimation method is **consistent**. Now we are ready to apply the estimator to real empirical data.

Benoit Mandelbrot noticed that you can make Zipf's Law fit the word frequency data better by adding a parameter $\beta$, to form the Zipf-Mandelbrot distribution:
$$
p(x) = \frac{1}{Z} \left(k+\beta\right)^{-\alpha}.
$$

Recall, 
$$
p(w) = \frac{1}{Z} k^{-\alpha},
$$
where:
* $Z$ is the normalizing constant that makes the distribution sum to 1.
* $k$ is the frequency rank of the word $w$. If $w$ is the most frequent word, it has frequency rank $k=1$, etc.
* $\alpha$ is a nonnegative number.

The Zipf distribution is implemented below for a vocabulary of size $N$:

**Exercise** Write the function for the Zipf-Mandelbrot distribution. Visualize it on a log-log plot. What is the effect of the parameter $\beta$?

```python
def zipf_mandelbrot_distribution(N, alpha, beta):
  k = np.arange(N).astype(float) + 1 # an array of [1, 2, ..., N]
  kb = (k + beta) 
  p = kb ** -alpha
  Z = p.sum()
  return p / Z
```

Q. Does this mean that the ZM model is a better model of the true underlying distribution?

What is going on when we compare cross entropies? 

> Measuring surprisal + reminds me of KL Divergence

Consider the difference of cross entropies for the Zipf model $q_Z$ and the ZM model $q_{ZM}$, applying our logarithm rules:

$$
H[p, q_{Z}] - H[p, q_{ZM}] = -\sum_w p(w) \log q_Z(w) + \sum_w p(w) \log q_{ZM}(w)
$$
$$
= \sum_w p(w) \left[\log q_{ZM}(w) - \log q_Z(w) \right]
$$
$$
= \sum_w p(w) \log \frac{q_{ZM}(w)}{q_Z(w)}.
$$

The model comparison comes out to a **log likelihood ratio** $\log \frac{q_{ZM}(w)}{q_Z(w)}$. In general, log likelihood ratio is how you should compare how well two models fit the data. **Likelihood** means the probability assigned to data by a model, such as $q_Z$ or $q_{ZM}$ in this case.
