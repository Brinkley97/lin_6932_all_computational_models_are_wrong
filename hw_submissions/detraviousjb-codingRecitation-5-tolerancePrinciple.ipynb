## The Tolerance Principle at a high level and a very high level

 Virtually all potential rules of a language are bound to have (arbitrary) exceptions, so the learner shouldn't hold rules to a standard of perfection. For example, a clear rule of English, suffix *-ed* to the verb to make past tense has plenty of exceptions. The real question is how many exceptions, as a proportion of the total number of types that are expected to follow the rule, can be tolerated. If there are “too many” exceptions, the rule is probably not a good one for the learner to acquire. If there aren’t “too many,” the learner should acquire the rule. The Tolerance Principle provides this threshold: If the rule should apply to $N$ types that the child knows so far, but the child has learned $e$ exceptions to the rule, the child should learn the rule + memorize the exceptions if $e$ < $\theta$, the tolerance threshold. If $e$ > $\theta$, it’s back to the drawing board. Either just memorize it all as exceptions, or try again with a different rule.

One one hand, language is incredibly arbitrary. On the other, it is ex-traordinarily rule-governed. This duality is what has secured linguistics its privileged role in the cognitive sciences. That said, it also presents an enormous challenge to language learners, who each have to work out for themselves what is productive/rule-governed and what is non- productive/arbitrary. The Tolerance Principle is a precise metric which the learner can use to make this decision.

**TP**: $\theta = \frac{N}{lnN}$

**The question**: Is $e < \theta$ ?

## Exercise 1: Performing some Calculations




1.   Say the learner hypothesizes some rule for which $N = 40$ and $e = 10$. Is the rule productive? Show your work by calculating $\theta$. 

> 10

```python
# Insert your code here
def calc_theta(N: int, e: int):
    theta = N / math.log(N)
    if isinstance(e, int):
        if e <= theta:
            print(f"{e} <= {theta}, so there aren't 'too many,' the learner should acquire the rule.")  
        else:
            print(f"{e} > {theta}, so there are 'too many' exceptions, the rule is probably not a good one for the learner to acquire.")
    return theta
```

```python
calc_theta(N=40, e=10)
```

4. If there are $e = 34$ exceptions to a rule, at least how large must $N$ be for them to be tolerable?

### Strong, Weak, Semi-Weak, and Suppletive

Four kinds of verbs are traditionally distinguished in traditional Germanic linguistics, strong verbs, weak verbs, semi-weak verbs, and suppletive verbs. 

1. stem: [[Lexicon of linguistics](https://lexicon.hum.uu.nl/) by Johan Kerstens, Eddy Ruys, Joost Zwarts]
    - MORPHOLOGY: a term which is commonly used for the **uninflected** part of a word. 
        - inflected [dynamically changes to encode grammatical information such as tense, person, or number]
        - uninflected [static or fixed regardless of tense, person, or subject agreement]
    - EXAMPLE: if we take the plural form disagreements, the form disagreement is called the stem.
        - inflected: -s
        - uninflected: disagreement
2. suffix: [[Lexicon of linguistics](https://lexicon.hum.uu.nl/) by Johan Kerstens, Eddy Ruys, Joost Zwarts]
    - MORPHOLOGY: a bound morpheme (or affix) which attaches at the righthand side of a base. 
    - EXAMPLE: the English morpheme -ness is a suffix, since it attaches to the right of adjectives (productiveness). 
        - inflected (of disagreements): -s

---

1. Strong verbs are ones that undergo a stem vowel change instead of suffixing to form pasts, like *sing*, *bite*, or *ride*.
    - sing: stem is sing + past form is sang,  $ \therefore $ the vowel change (i $ \rightarrow $ a).
    - bite: stem is bite + past form is bit.
    - ride: stem is ride + past form is rode,  $ \therefore $ the vowel change (i $ \rightarrow $ o).
2. Weak verbs are the ones that make pasts in -*ed* in English without any kind of stem change. These are regular in English.
    - predict: stem is predict + past form is predicted, $ \therefore $ the stem is without change + suffix change (-*ed*).
3. Semi-weak verbs both undergo stem change and suffixing a coronal obstruent (-*t* or -*d*) like sleep or tell. 
    - sleep: stem is sleep + past form is slept, $ \therefore $ the stem changes + suffix change (-*t*).
    - tell: stem is tell + past form is told, $ \therefore $ the stem changes + suffix change (-*d*).
4. Finally, suppletives are just straight-up irregulars like *go*.
    - go: stem is go + past form is went, $ \therefore $ the stem changes completely.

The terminology actually goes back to the Brothers Grimm. Mainly famous for their fairy tales, they were also foundational historical linguists and comparative Germanicists.

## Weak x Strong Verbs

```python
# df.loc[:, 'POS x Word x Tense']
pos_tags = []
words = []
tenses = []

for idx, row in df.iterrows():
    raw_pos_word_tense = row['POS x Word x Tense']
    print(f"Raw ({idx}): {raw_pos_word_tense}")

    separate_pos = raw_pos_word_tense.split('|')
    print(f" -> Separate: {separate_pos}")

    pos = separate_pos[0]
    print(f"  -> pos: {pos}")
    pos_tags.append(pos)

    word_and_tense = separate_pos[1]
    separator = '&' if '&' in word_and_tense else '-'
    separate_word_and_tense = word_and_tense.split(separator)
    word = separate_word_and_tense[0]
    tense = separate_word_and_tense[1]
    print(f"  -> word: {word} \n  -> tense {tense}")
    words.append(word)
    tenses.append(tense)
    print()

df['POS'] = pos_tags
df['Words'] = words
df['Tenses'] = tenses

df
```

```python
words = df.loc[:, 'Words'].to_list()
words
```

```python
from lemminflect import getInflection
get_past_tenses = []

for word in words:
    get_past_tense = getInflection(word, tag="VBD")[0]
    get_past_tenses.append(get_past_tense)

df['Past Tense Word'] = get_past_tenses
df
```

```python
past_tense_words = df.loc[:, 'Past Tense Word'].to_list()
past_tense_words
```
